### Target Guided Ordinal Encoding

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'city' : ['New York', 'Londan', 'Paris', "Tokyo", 'New York', 'Paris'],
    'price' : [100, 80, 90, 120, 110, 95]
})
```

```python
df
```

```python
mean_price_by_city = df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price_by_city
```

```python
df['city_encoded'] = df['city'].map(mean_price_by_city)
```
